# **Regular Expression**

```python
a=nltk.corpus.words.words
```

```python
wordlist=[w for w in a('en')]
```

```python
wordlist
```

```python
[w for w in wordlist if re.search("best$",w)]
```

```python
[w for w in wordlist if re.search('...a..n..s',w)]
```

```python
[w for w in wordlist if re.search('p+i+e$',w)]
```

# **Normalization**

```python
print(tokens)
```

```python
porter=nltk.PorterStemmer()
```

```python
lancaster=nltk.LancasterStemmer()
```

```python
[porter.stem(t) for t in tokens]
```

```python
[lancaster.stem(t) for t in tokens]
```

```python
wnl = nltk.WordNetLemmatizer()
[wnl.lemmatize(t) for t in tokens]
```
